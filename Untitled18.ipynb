```python
merged_df = merged_df.sort_values(['service', 'year', 'week'])

lag_cols = [
    'patients_request',
    'bed_utilization',
    'avg_length_of_stay',
    'demand_to_capacity_ratio'
]

for col in lag_cols:
    merged_df[f'{col}_lag_1'] = merged_df.groupby('service')[col].shift(1)
    merged_df[f'{col}_lag_4'] = merged_df.groupby('service')[col].shift(4)
```

```python
merged_df['requests_rolling_4w'] = (
    merged_df.groupby('service')['patients_request']
    .rolling(4)
    .mean()
    .reset_index(level=0, drop=True)
)
```

```python
merged_df['week_sin'] = np.sin(2 * np.pi * merged_df['week'] / 52)
merged_df['week_cos'] = np.cos(2 * np.pi * merged_df['week'] / 52)
```

```python
target = 'patients_request'
features = [c for c in merged_df.columns if c not in [
    'date_approx', 'event', target
]]
```

```python
for lag in [1, 2, 3, 4, 8, 12]:
    merged_df[f'patients_request_lag_{lag}'] = merged_df.groupby('service')['patients_request'].shift(lag)
    merged_df[f'refusal_rate_lag_{lag}'] = merged_df.groupby('service')['refusal_rate'].shift(lag)
    merged_df[f'bed_utilization_lag_{lag}'] = merged_df.groupby('service')['bed_utilization'].shift(lag)

# Lag differences (rate of change)
merged_df['request_change_1w'] = merged_df['patients_request'] - merged_df['patients_request_lag_1']
merged_df['request_change_4w'] = merged_df['patients_request'] - merged_df['patients_request_lag_4']
merged_df['request_pct_change_1w'] = (merged_df['request_change_1w'] / merged_df['patients_request_lag_1']) * 100

# Multiple rolling windows
for window in [2, 4, 8, 12]:
    merged_df[f'requests_rolling_{window}w'] = merged_df.groupby('service')['patients_request'].rolling(window).mean().reset_index(level=0, drop=True)
    merged_df[f'requests_rolling_{window}w_std'] = merged_df.groupby('service')['patients_request'].rolling(window).std().reset_index(level=0, drop=True)
    merged_df[f'requests_rolling_{window}w_max'] = merged_df.groupby('service')['patients_request'].rolling(window).max().reset_index(level=0, drop=True)
    merged_df[f'requests_rolling_{window}w_min'] = merged_df.groupby('service')['patients_request'].rolling(window).min().reset_index(level=0, drop=True)

# Rolling ratios
merged_df['request_to_rolling_avg'] = merged_df['patients_request'] / merged_df['requests_rolling_4w']
merged_df['request_zscore_4w'] = (merged_df['patients_request'] - merged_df['requests_rolling_4w']) / merged_df['requests_rolling_4w_std']

# Month cyclical encoding
merged_df['month_sin'] = np.sin(2 * np.pi * merged_df['month'] / 12)
merged_df['month_cos'] = np.cos(2 * np.pi * merged_df['month'] / 12)

# Quarter encoding
merged_df['quarter'] = ((merged_df['month'] - 1) // 3) + 1
merged_df['quarter_sin'] = np.sin(2 * np.pi * merged_df['quarter'] / 4)
merged_df['quarter_cos'] = np.cos(2 * np.pi * merged_df['quarter'] / 4)

# Day of year (if available)
if 'day_of_year' in merged_df.columns:
    merged_df['day_sin'] = np.sin(2 * np.pi * merged_df['day_of_year'] / 365)
    merged_df['day_cos'] = np.cos(2 * np.pi * merged_df['day_of_year'] / 365)
```

```python
if 'event' in merged_df.columns:
    merged_df['has_event'] = (merged_df['event'] != 'none').astype(int)
    merged_df['event_flu'] = (merged_df['event'] == 'flu').astype(int)
    merged_df['event_holiday'] = (merged_df['event'].isin(['holiday', 'christmas', 'new_year'])).astype(int)
```

```python
merged_df['utilization_x_demand_ratio'] = merged_df['bed_utilization'] * merged_df['demand_to_capacity_ratio']
merged_df['request_x_utilization'] = merged_df['patients_request'] * merged_df['bed_utilization']
merged_df['lag1_x_lag4'] = merged_df['patients_request_lag_1'] * merged_df['patients_request_lag_4']

# Ratio features
merged_df['request_to_capacity'] = merged_df['patients_request'] / merged_df['available_beds']
merged_df['utilization_to_refusal'] = merged_df['bed_utilization'] / (merged_df['refusal_rate'] + 1)
```

```python
from sklearn.preprocessing import PolynomialFeatures

poly_features = ['bed_utilization', 'demand_to_capacity_ratio', 'avg_length_of_stay']
poly_transformer = PolynomialFeatures(degree=2, include_bias=False, interaction_only=True)
poly_data = poly_transformer.fit_transform(merged_df[poly_features])
poly_df = pd.DataFrame(poly_data, columns=poly_transformer.get_feature_names_out(poly_features))
merged_df = pd.concat([merged_df, poly_df], axis=1)
```

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, mean_absolute_percentage_error

target = 'patients_request'

# Redefine X to include all engineered features from merged_df
columns_to_drop = ['date_approx', 'event', target, 'service']
X = merged_df.drop(columns=columns_to_drop, errors='ignore').replace([np.inf, -np.inf], np.nan)

# Explicitly convert all columns to numeric (float) and fill any remaining NaNs
# Using .apply() ensures pd.to_numeric is called on a Series for each column
X = X.apply(pd.to_numeric, errors='coerce')
X = X.fillna(X.median())

# Update the features list to reflect the actual columns in X
features = X.columns.tolist()
y = merged_df[target]

# Split data (temporal split)
split_point = int(len(merged_df) * 0.8)
X_train, X_test = X.iloc[:split_point], X.iloc[split_point:]
y_train, y_test = y.iloc[:split_point], y.iloc[split_point:]

# Train Gradient Boosting
gb_model = GradientBoostingRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=4,
    min_samples_leaf=20,
    random_state=42
)

# Drop NaNs from training data to avoid issues, as fillna might leave some if entire column is NaN
X_train_processed = X_train.dropna()
y_train_processed = y_train.loc[X_train_processed.index]

gb_model.fit(X_train_processed, y_train_processed)

# Evaluate
X_test_processed = X_test.dropna() # Drop NaNs from test data for evaluation
y_test_clean = y_test.loc[X_test_processed.index]
y_pred = gb_model.predict(X_test_processed)

mae = mean_absolute_error(y_test_clean, y_pred)
rmse = np.sqrt(mean_squared_error(y_test_clean, y_pred))
r2 = r2_score(y_test_clean, y_pred)
mape = mean_absolute_percentage_error(y_test_clean, y_pred) * 100

print("=" * 60)
print("GRADIENT BOOSTING MODEL RESULTS")
print("=" * 60)
print(f"MAE:  {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R²:   {r2:.3f}")
print(f"MAPE: {mape:.2f}%")

# ============================================================================
# CELL 2: ADD XGBOOST MODEL
# ============================================================================

from xgboost import XGBRegressor

xgb_model = XGBRegressor(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1,
    verbosity=0
)

# Convert to numpy arrays before fitting XGBoost
X_train_processed_np = X_train_processed.astype(float).values
y_train_processed_np = y_train_processed.astype(float).values

xgb_model.fit(X_train_processed_np, y_train_processed_np)

# Convert test data to numpy arrays for prediction
X_test_processed_np = X_test_processed.astype(float).values
xgb_pred = xgb_model.predict(X_test_processed_np)

xgb_mae = mean_absolute_error(y_test_clean, xgb_pred)
xgb_rmse = np.sqrt(mean_squared_error(y_test_clean, xgb_pred))
xgb_r2 = r2_score(y_test_clean, xgb_pred)
xgb_mape = mean_absolute_percentage_error(y_test_clean, xgb_pred) * 100

print("=" * 60)
print("XGBOOST MODEL RESULTS")
print("=" * 60)
print(f"MAE:  {xgb_mae:.2f}")
print(f"RMSE: {xgb_rmse:.2f}")
print(f"R²:   {xgb_r2:.3f}")
print(f"MAPE: {xgb_mape:.2f}%")

# Compare models
print("\n" + "=" * 60)
print("MODEL COMPARISON")
print("=" * 60)
comparison = pd.DataFrame({
    'Model': ['Gradient Boosting', 'XGBoost'],
    'MAE': [mae, xgb_mae],
    'RMSE': [rmse, xgb_rmse],
    'R²': [r2, xgb_r2],
    'MAPE': [mape, xgb_mape]
})
print(comparison.to_string(index=False))

# Select best model
best_model = gb_model if mae < xgb_mae else xgb_model
best_model_name = 'Gradient Boosting' if mae < xgb_mae else 'XGBoost'
print(f"\n✓ Best Model: {best_model_name}")
```

```python
# Generate forecasts for each service
forecasts = []

# Get the list of features the model was trained on
# This should be X_train_processed.columns
model_trained_features = X_train_processed.columns.tolist()

for service in merged_df['service'].unique():
    service_data = merged_df[merged_df['service'] == service].sort_values(['year', 'week']).tail(1).copy()

    if len(service_data) > 0:
        # Create a new DataFrame for last_features that has exactly the same columns
        # as the training data, initialized from service_data.
        # This handles cases where service_data might have extra columns
        # or where X_train_processed might have had fewer columns due to drops.
        last_features = pd.DataFrame(columns=model_trained_features, index=service_data.index)
        for col in model_trained_features:
            if col in service_data.columns:
                last_features[col] = service_data[col]
            else:
                # If a feature from training is not in service_data, fill with median from training
                # Ensure the column exists in X_train_processed before attempting to get its median
                if col in X_train_processed.columns:
                    last_features[col] = X_train_processed[col].median() # Use median of training data
                else:
                    last_features[col] = 0 # Fallback for new features not in training set, if any

        # Handle NaNs and Infs BEFORE updating temporal features, to ensure values are valid for lag calculations
        # Ensure all columns are numeric
        last_features = last_features.apply(pd.to_numeric, errors='coerce')
        last_features = last_features.fillna(X_train_processed.median()) # Fill with median from training data for consistency
        last_features = last_features.replace([np.inf, -np.inf], np.nan).fillna(X_train_processed.median())


        current_week = merged_df['week'].max()

        # Forecast 4 weeks ahead
        for week_ahead in range(1, 5):
            future_week = current_week + week_ahead
            # Calculate month and quarter, ensuring they don't exceed 12 and 4 respectively
            future_month = (future_week - 1) % 12 + 1
            future_quarter = (future_month - 1) // 3 + 1

            # Update temporal features - ensure these operations don't add new columns or change column order
            if 'week' in last_features.columns:
                last_features['week'] = future_week
            if 'month' in last_features.columns:
                last_features['month'] = future_month
            if 'quarter' in last_features.columns:
                last_features['quarter'] = future_quarter
            if 'week_sin' in last_features.columns:
                last_features['week_sin'] = np.sin(2 * np.pi * future_week / 52)
            if 'week_cos' in last_features.columns:
                last_features['week_cos'] = np.cos(2 * np.pi * future_week / 52)
            if 'month_sin' in last_features.columns:
                last_features['month_sin'] = np.sin(2 * np.pi * future_month / 12)
            if 'month_cos' in last_features.columns:
                last_features['month_cos'] = np.cos(2 * np.pi * future_month / 12)
            if 'quarter_sin' in last_features.columns:
                last_features['quarter_sin'] = np.sin(2 * np.pi * future_quarter / 4)
            if 'quarter_cos' in last_features.columns:
                last_features['quarter_cos'] = np.cos(2 * np.pi * future_quarter / 4)

            # Convert to numpy array for prediction
            last_features_np = last_features.astype(float).values

            # Predict
            pred = best_model.predict(last_features_np)[0]

            forecasts.append({
                'service': service,
                'week': future_week,
                'forecast': max(0, int(pred))
            })

            # Update lag features for next prediction iteration within the same service
            # This assumes that 'patients_request_lag_1' should become the predicted value
            # and other lags shift down.
            if 'patients_request_lag_1' in last_features.columns:
                # Shift existing lag values before assigning the new prediction
                # Iterate down from the largest lag (e.g., 12) to ensure correct shifting
                for lag_val in [12, 8, 4, 3, 2]: # Adjusted to include relevant lags based on feature engineering
                    if f'patients_request_lag_{lag_val}' in last_features.columns:
                        prev_lag_col = f'patients_request_lag_{lag_val-1}'
                        if lag_val == 1:
                            # This case should be handled by the direct assignment below
                            continue
                        elif prev_lag_col in last_features.columns:
                            last_features[f'patients_request_lag_{lag_val}'] = last_features[prev_lag_col].values[0]
                        else:
                            # If the previous lag column doesn't exist, fill with median from training
                            if f'patients_request_lag_{lag_val}' in X_train_processed.columns:
                                last_features[f'patients_request_lag_{lag_val}'] = X_train_processed[f'patients_request_lag_{lag_val}'].median()
                            else:
                                last_features[f'patients_request_lag_{lag_val}'] = np.nan # Or a default value
                last_features['patients_request_lag_1'] = pred # Update the immediate lag with the new prediction

                # Apply similar logic for other lagged variables if they exist in last_features
                for col_prefix in ['refusal_rate_lag_', 'bed_utilization_lag_']:
                    for lag_val in [12, 8, 4, 3, 2]:
                        if f'{col_prefix}{lag_val}' in last_features.columns:
                            prev_lag_col = f'{col_prefix}{lag_val-1}'
                            if lag_val == 1:
                                continue # Handled by specific prediction for request_lag_1
                            elif prev_lag_col in last_features.columns:
                                last_features[f'{col_prefix}{lag_val}'] = last_features[prev_lag_col].values[0]
                            else:
                                if f'{col_prefix}{lag_val}' in X_train_processed.columns:
                                    last_features[f'{col_prefix}{lag_val}'] = X_train_processed[f'{col_prefix}{lag_val}'].median()
                                else:
                                    last_features[f'{col_prefix}{lag_val}'] = np.nan

forecast_df = pd.DataFrame(forecasts)
forecast_pivot = forecast_df.pivot(index='week', columns='service', values='forecast')

print("=" * 60)
print("4-WEEK FORECAST BY SERVICE")
print("=" * 60)
print(forecast_pivot.to_string())

# Visualize forecasts
fig, ax = plt.subplots(figsize=(12, 6))
for service in forecast_df['service'].unique():
    service_forecast = forecast_df[forecast_df['service'] == service]
    ax.plot(service_forecast['week'], service_forecast['forecast'],
            marker='o', linewidth=2, label=service)

ax.set_xlabel('Week')
ax.set_ylabel('Forecasted Requests')
ax.set_title('4-Week Forecast by Service', fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```
